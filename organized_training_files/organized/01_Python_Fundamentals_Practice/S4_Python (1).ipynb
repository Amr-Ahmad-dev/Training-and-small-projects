Managing a Simple Inventory System

```python
class inventory:
    def __init__(self):
        self.items = {}

    def add_item(self, item_name, quantity):
        if item_name in self.items:
            self.items[item_name] += quantity
        else:
            self.items[item_name] = quantity

    def remove_item(self, item_name, quantity):
        if item_name in self.items and self.items[item_name] >= quantity:
            self.items[item_name] -= quantity
            if self.items[item_name] == 0:
                del self.items[item_name]
        else:
            print(f"Cannot remove {quantity} of {item_name}. Not enough in inventory.")

    def get_inventory(self):
        return self.items
```

Modeling a Bank Account

```python
class bank_account:
    def __init__(self, account_number, account_holder, balance=0):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
        else:
            print("Insufficient funds or invalid withdrawal amount.")

    def get_balance(self):
        return self.balance
```

Modeling a Simple Library System

```python
class library:
    def __init__(self):
        self.books = {}

    def add_book(self, title, author, quantity):
        if title in self.books:
            self.books[title]['quantity'] += quantity
        else:
            self.books[title] = {'author': author, 'quantity': quantity}

    def remove_book(self, title, quantity):
        if title in self.books and self.books[title]['quantity'] >= quantity:
            self.books[title]['quantity'] -= quantity
            if self.books[title]['quantity'] == 0:
                del self.books[title]
        else:
            print(f"Cannot remove {quantity} of {title}. Not enough in library.")

    def get_books(self):
        return self.books   
```

Simple Voting System

```python
class Tideman:
    def __init__(self, candidates):
        self.candidates = candidates
        self.num_candidates = len(candidates)
        self.preferences = [[0] * self.num_candidates for _ in range(self.num_candidates)]
        self.locked = [[False] * self.num_candidates for _ in range(self.num_candidates)]
        self.pairs = []
    def record_preferences(self, ranks):
        for i in range(len(ranks)):
            for j in range(i + 1, len(ranks)):
                preferred = ranks[i]
                over = ranks[j]
                self.preferences[preferred][over] += 1
    def add_pairs(self):
        for i in range(self.num_candidates):
            for j in range(i + 1, self.num_candidates):
                if self.preferences[i][j] > self.preferences[j][i]:
                    self.pairs.append((i, j))
                elif self.preferences[j][i] > self.preferences[i][j]:
                    self.pairs.append((j, i))
    def sort_pairs(self):

        self.pairs.sort(
            key=lambda pair: self.preferences[pair[0]][pair[1]] - self.preferences[pair[1]][pair[0]],
            reverse=True
        )
    def _has_cycle(self, winner, loser):

        if loser == winner:
            return True

        for candidate in range(self.num_candidates):
            if self.locked[loser][candidate]:
                if self._has_cycle(winner, candidate):
                    return True
        return False

    def lock_pairs(self):
 
        for winner, loser in self.pairs:
            if not self._has_cycle(winner, loser):
                self.locked[winner][loser] = True

    def print_winner(self):
        for candidate in range(self.num_candidates):
            has_incoming_edge = False
            for other in range(self.num_candidates):
                if self.locked[other][candidate]:
                    has_incoming_edge = True
                    break
            
            if not has_incoming_edge:
                return self.candidates[candidate]
        return None


"""yesssssssssssssssssssssssss i  did it """
```

Simple Temperature Converter

```python
class temp_conv:
    def __init__(self,type):
        self.type = type
    def convert(self, value):
        if self.type == "CtoF":
            return (value * 9/5) + 32
        elif self.type == "FtoC":
            return (value - 32) * 5/9
        else:
            raise ValueError("Invalid conversion type. Use 'CtoF' or 'FtoC'.")
```

Modeling a Car Rental System

```python
class car_rent:
    def __init__(self, car_type, daily_rate):
        self.car_type = car_type
        self.daily_rate = daily_rate

    def calculate_rental_cost(self, days):
        if days <= 0:
            raise ValueError("Number of rental days must be positive.")
        return self.daily_rate * days
```
