Finding Prime Numbers within a Range

```python
import math

range_num = int(input("int num: "))

all_p = []

for hard in range(2, range_num + 1):
    p = 1
    for com_on_man in range(2, int(math.sqrt(hard)) + 1):
        if hard % com_on_man == 0:
            p = 0
            break

    if p == 1:
        all_p.append(hard)

print(all_p)
```

Generating a Fibonacci Sequence

```python
int_r=int(input("range"))
n_1=1
n_2=1
sequence=[1,1]
for i in range(int_r-2):
    sequence.append(n_1+n_2)
    n_1=n_2
    n_2=sequence[-1]

print(sequence)
```

Finding Common Elements in Two Lists

```python
l_1=[1,2,3,4,5,55,5,55,5,55]
l_2=[1,2,34,5,5658,9,]
comen=[]
for i in range(len(l_1)):
    for j in range(len(l_2)):
        if l_1[i]==l_2[j]:
            if l_1[i] not in comen:
                comen.append(l_1[i])

print(comen)
```

Calculating Factorial of a Number

```python
i_m=int(input("num: "))
num=i_m
while True:
    if i_m==1:
        break
    i_m-=1
    num*=i_m
print(num)
```

Checking for Palindrome Strings

```python
text = input("Enter a string: ")

is_palindrome = True

for i in range(int(len(text) / 2)):
    if text[i] != text[-i - 1]:
        is_palindrome = False
        break

if is_palindrome:
    print("Palindrome")
else:
    print("Not a palindrome")
```
